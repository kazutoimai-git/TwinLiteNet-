```python
import torch
from argparse import ArgumentParser
import yaml
from ptflops import get_model_complexity_info

from model.model import TwinLiteNetPlus  # ← あなたのモデル


def build_args():
    """Train と同じ引数構造を再現"""
    parser = ArgumentParser()
    parser.add_argument('--config', default='nano')
    parser.add_argument('--ema', action='store_true')
    parser.add_argument('--batch_size', type=int, default=16)
    parser.add_argument('--num_workers', type=int, default=16)
    parser.add_argument('--savedir', default='./result_v2/base_proposed/nano')
    parser.add_argument('--hyp', type=str, default='./hyperparameters/twinlitev2_hyper.yaml')
    parser.add_argument('--resume', type=str, default='')
    parser.add_argument('--verbose', action='store_true')
    
    args = parser.parse_args([])
    return args


if __name__ == "__main__":
    # ----------- args と hyp をロード ----------
    args = build_args()

    with open(args.hyp, errors='ignore') as f:
        hyp = yaml.safe_load(f)

    # ----------- モデル構築 ----------
    model = TwinLiteNetPlus(args)
    model.eval()

    # 入力解像度（例：3×360×640）
    input_res = (3, hyp.get('h', 360), hyp.get('w', 640))

    # ----------- FLOPs & Params 計算 ----------
    flops, params = get_model_complexity_info(
        model,
        input_res,
        as_strings=True,
        print_per_layer_stat=False,
        verbose=False
    )

    print("==================================")
    print(" Model Complexity (ptflops) ")
    print("==================================")
    print(f"Input resolution: {input_res}")
    print(f"FLOPs:   {flops}")
    print(f"Params:  {params}")
    print("==================================")
```
